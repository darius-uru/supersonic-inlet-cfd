```python
import numpy as np
import matplotlib.pyplot as plt

# Problem 3
# Eulerian velocity field:
# v = x1 e1 - x2 e2

# Spatial domain: first quadrant
x1 = np.linspace(0, 5, 25)
x2 = np.linspace(0, 5, 25)

X1, X2 = np.meshgrid(x1, x2)

# Velocity components
V1 = X1
V2 = -X2

# Plot
plt.figure(figsize=(8, 5))

# Velocity field
plt.quiver(X1, X2, V1, V2, alpha=0.5, color = "navy",label = "Velocity Field")

# Streamlines
plt.streamplot(
    X1, X2,
    V1, V2,
    density=1.0,
    linewidth=1.2,
    arrowsize=1.2, color = "Navy"

)

# Stagnation point
plt.scatter(0, 0, s=60, zorder=5, color = 'black', label="Stagnation point")

# Axes / walls
plt.axhline(y = 0, xmin=0,xmax = 5, linewidth=2, color = "red")
plt.axvline(x=0,ymin = 0, ymax = 5, linewidth=2, color = "red", label = "BC")

plt.xlabel(r"$x_1$")
plt.ylabel(r"$x_2$")
plt.title(r"Velocity Field and Streamlines: $\mathbf{v}=x_1\hat{e}_1-x_2\hat{e}_2$")
plt.axis("equal")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
